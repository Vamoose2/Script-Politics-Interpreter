## Script Politics Interpreter
This is a project which aims to take each line within a TV Episode or film and identify the overall political sentiment of it. This does not look at individual characters, rather a more holistic overview of the script. The code should be able to list the most left leaning and right leaning lines of the script.

### Dataset
The dataset used in this version includes 274 political statements genetated by an LLM. Half being left-leaning, and half being right-leaning. This dataset is far from perfect, but for this project, it will be sufficient to allow for exploration of NLP techniques.
